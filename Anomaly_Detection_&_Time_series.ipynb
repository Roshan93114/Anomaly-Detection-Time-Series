#1. What is Anomaly Detection? Explain its types (point, contextual, and
collective anomalies) with examples.
Answer:Anomaly Detection is the process of identifying data points, patterns, or observations that deviate significantly from normal or expected behavior in a dataset. These unusual patterns are called anomalies (or outliers) and often indicate important events such as fraud, system failures, intrusions, or errors.

Types of Anomalies
1. Point Anomaly

A point anomaly occurs when a single data instance is clearly different from the rest of the data.

Explanation:
One specific value stands out as abnormal when compared to the majority of observations.

Example:

In a dataset of daily bank withdrawals, most transactions are under ₹10,000, but one transaction is ₹5,00,000.

In temperature readings where values are around 25–30°C, a single reading of 70°C is a point anomaly.

2. Contextual Anomaly

A contextual anomaly is abnormal only within a specific context, such as time, location, or season.

Explanation:
The same data point may be normal in one context but anomalous in another.

Example:

A temperature of 30°C is normal in summer but anomalous in winter.

High website traffic at 12 PM may be normal, but the same traffic at 3 AM could be anomalous.

Spending ₹20,000 may be normal for a holiday season but unusual on a regular day.

3. Collective Anomaly

A collective anomaly occurs when a group of related data points is anomalous, even though individual points may appear normal.

Explanation:
The abnormality is detected only when looking at the pattern or sequence as a whole.

Example:

A sudden sequence of small but frequent network requests indicating a DDoS attack.

A series of normal-looking sensor readings that together indicate equipment malfunction.

Continuous small withdrawals from an account that collectively suggest fraudulent activity.

Summary Table
Type of Anomaly	Description	Example
Point	Single data point is abnormal	One very large bank transaction
Contextual	Abnormal in a specific context	High temperature in winter
Collective	Group of data points is abnormal	Repeated small network requests


#2.Compare Isolation Forest, DBSCAN, and Local Outlier Factor in terms of
their approach and suitable use cases.
Answer:Below is a clear comparison of Isolation Forest, DBSCAN, and Local Outlier Factor (LOF) focusing on their approach and suitable use cases.

1. Isolation Forest (iForest)
Approach

Tree-based anomaly detection method.

Works by randomly partitioning data using decision trees.

Anomalies are isolated faster because they require fewer splits than normal points.

Does not rely on distance or density measures.

Key Characteristics

Efficient for large and high-dimensional datasets

Handles non-linear relationships well

Requires setting the number of trees and sample size

Suitable Use Cases

Fraud detection in financial transactions

Intrusion detection in cybersecurity

High-dimensional data (e.g., log data, sensor data)

When anomalies are rare and different from normal data

2. DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
Approach

Density-based clustering algorithm.

Groups points that are closely packed together.

Points in low-density regions are labeled as anomalies (noise).

Uses two parameters:

ε (epsilon): neighborhood radius

MinPts: minimum points to form a dense region

Key Characteristics

Can detect arbitrarily shaped clusters

No need to predefine the number of clusters

Sensitive to parameter selection

Struggles with varying densities and high-dimensional data

Suitable Use Cases

Spatial data analysis

Geographical data (e.g., GPS coordinates)

Image processing

Anomaly detection when clusters are well-defined

3. Local Outlier Factor (LOF)
Approach

Density-based local anomaly detection technique.

Compares the local density of a point with that of its neighbors.

Points with significantly lower density than neighbors are anomalies.

Key Characteristics

Effective at detecting local anomalies

Requires choosing the number of neighbors (k)

Computationally expensive for large datasets

Sensitive to noise and parameter choice

Suitable Use Cases

Detecting anomalies in datasets with varying density

Network intrusion detection

Fraud detection where anomalies are subtle and context-dependent

Small to medium-sized datasets

#3. What are the key components of a Time Series? Explain each with one
example.
Answer:A Time Series is a sequence of data points recorded at regular time intervals. Time series analysis helps understand patterns over time and make forecasts.

Key Components of a Time Series
1. Trend

Definition:
The long-term movement or direction of the data over time (upward, downward, or constant).

Example:

The steady increase in annual population of a country over several years.

A gradual rise in monthly online sales over time.

2. Seasonality

Definition:
Regular, repeating patterns that occur at fixed intervals (daily, monthly, yearly).

Example:

Higher sales of ice cream during summer every year.

Increased electricity usage during evenings each day.

3. Cyclical Component

Definition:
Fluctuations that occur over longer periods and are influenced by economic or business cycles, without a fixed duration.

Example:

Changes in unemployment rates due to economic booms and recessions.

Rise and fall of stock market performance over several years.

4. Irregular (Random) Component

Definition:
Unpredictable, random variations caused by unexpected events.

Example:

A sudden drop in airline bookings due to natural disasters.

Unexpected spikes in product demand due to viral trends.

Summary Table
Component	Meaning	Example
Trend	Long-term direction	Rising online sales
Seasonality	Repeating short-term pattern	Summer ice-cream sales
Cyclical	Long-term business cycles	Economic recessions
Irregular	Random, unpredictable effects	Sudden demand spikes  


#4. Define Stationary in time series. How can you test and transform a
non-stationary series into a stationary one?
Answer:Stationarity in Time Series refers to a property where the statistical characteristics of a time series remain constant over time. Many time-series models (like ARIMA) assume stationarity because it makes patterns easier to model and predict.

Definition of Stationary Time Series

A time series is said to be stationary if:

Mean is constant over time

Variance is constant over time

Covariance between two time points depends only on the lag, not on time itself

If any of these change over time, the series is non-stationary.

Example:

Daily temperature fluctuations around a constant average → Stationary

Increasing monthly sales over years → Non-stationary (changing mean)

Testing for Stationarity
1. Visual Inspection

Plot the time series

Look for trends, seasonality, or changing variance

Example:
An upward sloping graph indicates non-stationarity.

2. Statistical Tests
(a) Augmented Dickey-Fuller (ADF) Test

Null hypothesis (H₀): Series is non-stationary

If p-value < 0.05 → Reject H₀ → Stationary

(b) KPSS Test

Null hypothesis (H₀): Series is stationary

If p-value < 0.05 → Non-stationary

3. Rolling Statistics

Calculate rolling mean and variance

If they change over time → non-stationary

#6.Load a time series dataset (e.g., AirPassengers), plot the original series,
and decompose it into trend, seasonality, and residual components
Answer:Below is a complete example using the AirPassengers dataset that shows how to load the dataset, plot the original time series, and decompose it into trend, seasonality, and residual components.

Step 1: Import Required Libraries


```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

Step 2: Load the AirPassengers Dataset

```python
# Load dataset
data = pd.read_csv(
    "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv",
    parse_dates=["Month"],
    index_col="Month"
)

# Rename column for convenience
data.columns = ["Passengers"]

print(data.head())
```

Step 3: Plot the Original Time Series

```python
plt.figure(figsize=(10, 4))
plt.plot(data, label="Air Passengers")
plt.title("AirPassengers Time Series")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.legend()
plt.show()
```

Observation:
The series shows a clear upward trend and seasonal pattern, indicating it is non-stationary.

Step 4: Decompose the Time Series

We use additive decomposition (Trend + Seasonality + Residual).

```python
decomposition = seasonal_decompose(data, model="additive", period=12)

decomposition.plot()
plt.show()
```

Step 5: Explanation of Components
1. Trend

Shows the long-term increase in air passengers over time.

2. Seasonality

Repeating yearly pattern (higher passengers during certain months each year).

3. Residual (Irregular)

Random noise after removing trend and seasonality.

#7. Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to
detect anomalies. Visualize the anomalies on a 2D scatter plot.
Answer:-Below is a complete, practical example showing how to apply Isolation Forest on a numerical dataset (NYC Taxi Fare–like data) and visualize anomalies on a 2D scatter plot.

We will:

Load a numerical dataset

Apply Isolation Forest

Label anomalies

Visualize results

Step 1: Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
```

Step 2: Load NYC Taxi Fare–Style Dataset

(Using a public sample dataset with fare and trip distance)

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/taxis.csv"
data = pd.read_csv(url)

# Select numerical features
df = data[["distance", "fare"]].dropna()

df.head()
```

Step 3: Apply Isolation Forest

```python
# Initialize Isolation Forest
iso_forest = IsolationForest(
    n_estimators=100,
    contamination=0.05,
    random_state=42
)

# Fit and predict anomalies
df["anomaly"] = iso_forest.fit_predict(df)

# Convert labels: -1 → anomaly, 1 → normal
df["anomaly"] = df["anomaly"].map({1: 0, -1: 1})

df.head()
```

Step 4: Visualize Anomalies (2D Scatter Plot)

```python
plt.figure(figsize=(8, 6))

# Normal points
plt.scatter(
    df[df["anomaly"] == 0]["distance"],
    df[df["anomaly"] == 0]["fare"],
    c="blue",
    label="Normal",
    alpha=0.5
)

# Anomalies
plt.scatter(
    df[df["anomaly"] == 1]["distance"],
    df[df["anomaly"] == 1]["fare"],
    c="red",
    label="Anomaly",
    marker="x"
)

plt.xlabel("Trip Distance")
plt.ylabel("Fare Amount")
plt.title("Isolation Forest Anomaly Detection (NYC Taxi Fare)")
plt.legend()
plt.show()
```

Interpretation

Blue points: Normal taxi trips

Red crosses: Anomalies (e.g., unusually high fare for short distance or very long trips with low fare)

Isolation Forest isolates these rare and unusual patterns efficiently.

Why Isolation Forest Works Well Here

Handles large datasets

Works well with numerical features

Does not assume data distribution

Ideal for fraud detection and pricing anomalies

#8. Train a SARIMA model on the monthly airline passengers dataset.
Forecast the next 12 months and visualize the results.
Answer:-Below is a step-by-step example to train a SARIMA model on the monthly AirPassengers dataset, forecast the next 12 months, and visualize the results.

Step 1: Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
```

Step 2: Load the AirPassengers Dataset

```python
# Load dataset
data = pd.read_csv(
    "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv",
    parse_dates=["Month"],
    index_col="Month"
)

data.columns = ["Passengers"]
data.head()
```

Step 3: Train the SARIMA Model

We use:

(p, d, q) = (1, 1, 1)

Seasonal (P, D, Q, s) = (1, 1, 1, 12) for monthly seasonality

```python
model = SARIMAX(
    data["Passengers"],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

results = model.fit()
print(results.summary())
```

Step 4: Forecast the Next 12 Months

```python
forecast = results.get_forecast(steps=12)
forecast_mean = forecast.predicted_mean
confidence_intervals = forecast.conf_int()
```

Step 5: Visualize the Forecast

```python
plt.figure(figsize=(10, 5))

# Plot historical data
plt.plot(data.index, data["Passengers"], label="Observed")

# Plot forecast
plt.plot(forecast_mean.index, forecast_mean, label="Forecast", color="red")

# Plot confidence intervals
plt.fill_between(
    confidence_intervals.index,
    confidence_intervals.iloc[:, 0],
    confidence_intervals.iloc[:, 1],
    color="pink",
    alpha=0.3,
    label="Confidence Interval"
)

plt.title("SARIMA Forecast for AirPassengers (Next 12 Months)")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.legend()
plt.show()
```

Interpretation

The forecast line shows expected passenger counts for the next year.

The confidence interval reflects forecast uncertainty.

The model successfully captures both trend and seasonality.

#9. Apply Local Outlier Factor (LOF) on any numerical dataset to detect
anomalies and visualize them using matplotlib.
Answer:-Below is a complete example of applying Local Outlier Factor (LOF) on a numerical dataset and visualizing detected anomalies using matplotlib.

We will use a simple 2D numerical dataset so anomalies are easy to see.

Step 1: Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor
```

Step 2: Create a Numerical Dataset

We generate normal data points and add a few outliers.

```python
# Generate normal data
np.random.seed(42)
X_normal = np.random.normal(loc=0, scale=1, size=(300, 2))

# Generate anomalies
X_outliers = np.random.uniform(low=-6, high=6, size=(20, 2))

# Combine dataset
X = np.vstack((X_normal, X_outliers))
```

Step 3: Apply Local Outlier Factor (LOF)

```python
# Initialize LOF
lof = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.06
)

# Fit and predict
y_pred = lof.fit_predict(X)

# Convert labels: -1 = anomaly, 1 = normal
anomalies = X[y_pred == -1]
normal = X[y_pred == 1]
```

Step 4: Visualize Anomalies

```python
plt.figure(figsize=(8, 6))

# Plot normal points
plt.scatter(
    normal[:, 0],
    normal[:, 1],
    c="blue",
    label="Normal",
    alpha=0.6
)

# Plot anomalies
plt.scatter(
    anomalies[:, 0],
    anomalies[:, 1],
    c="red",
    label="Anomaly",
    marker="x",
    s=80
)

plt.title("Local Outlier Factor (LOF) Anomaly Detection")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```

Interpretation

Blue points: Normal observations

Red crosses: Detected anomalies

LOF identifies points with significantly lower local density than their neighbors.

Why LOF is Useful

Detects local anomalies

Works well when data has varying densities

Effective for fraud detection, network intrusion, and sensor data

#10.You are working as a data scientist for a power grid monitoring company.
Your goal is to forecast energy demand and also detect abnormal spikes or drops in
real-time consumption data collected every 15 minutes. The dataset includes features
like timestamp, region, weather conditions, and energy usage.
Explain your real-time data science workflow:
● How would you detect anomalies in this streaming data (Isolation Forest / LOF /
DBSCAN)?
● Which time series model would you use for short-term forecasting (ARIMA /
SARIMA / SARIMAX)?
● How would you validate and monitor the performance over time?
● How would this solution help business decisions or operations?

Answser:-Here’s a complete explanation of the workflow along with Python code and example outputs for real-time energy demand monitoring, anomaly detection, and short-term forecasting.

1. Real-Time Data Science Workflow
Step A: Anomaly Detection

Recommended Approach: Isolation Forest

Works well with high-dimensional streaming data (energy usage, weather, region)

Detects unusual spikes or drops without assuming data distribution

Alternative: LOF for local anomalies per region

Avoid DBSCAN for real-time because it is batch-oriented and computationally expensive

Workflow:

Maintain a sliding window of recent historical data

Train Isolation Forest on normal behavior

Score new data points every 15 minutes

Trigger alerts for anomalies (spikes/drops)

Step B: Short-Term Forecasting

Recommended Model: SARIMAX

Handles seasonality (daily/weekly cycles) and trend

Incorporates exogenous variables like temperature, humidity, holidays

Forecast Horizon: Next 1–24 hours (short-term)

Step C: Validation & Monitoring

Forecast Metrics: MAE, RMSE, MAPE

Anomaly Detection Metrics: False positives, true positives (compare with historical incidents)

Continuous Monitoring:

Data drift detection (changes in distribution)

Model retraining on sliding windows

Adaptive thresholds for alerts

Step D: Business & Operational Impact

Operational Benefits: Early detection of grid failures, proactive load balancing

Cost Savings: Reduced outages and penalties

Strategic Decisions: Regional monitoring, infrastructure planning, renewable integration

2. Python Implementation (Simulated Example)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from statsmodels.tsa.statespace.sarimax import SARIMAX

# ------------------------------
# Step 2: Simulate 15-min Energy Usage Data
# ------------------------------
np.random.seed(42)
date_rng = pd.date_range(start='2025-01-01', end='2025-01-02', freq='15min')  # 15-min intervals
n = len(date_rng)

# Simulated energy usage with daily seasonality
usage = (50 + 10*np.sin(2*np.pi*date_rng.hour/24) + np.random.normal(0, 2, n)).values # Convert to numpy array
usage[20] = 90   # spike anomaly
usage[60] = 10   # drop anomaly

# Simulate temperature as exogenous variable
temperature = 20 + 5*np.sin(2*np.pi*date_rng.hour/24) + np.random.normal(0, 1, n)

# Build DataFrame
df = pd.DataFrame({'timestamp': date_rng, 'energy_usage': usage, 'temperature': temperature})
df.set_index('timestamp', inplace=True)

# ------------------------------
# Step 3: Anomaly Detection with Isolation Forest
# ------------------------------
iso_forest = IsolationForest(contamination=0.05, random_state=42)
df['anomaly'] = iso_forest.fit_predict(df[['energy_usage', 'temperature']])
df['anomaly'] = df['anomaly'].map({1:0, -1:1})

# Plot anomalies
plt.figure(figsize=(12,5))
plt.plot(df.index, df['energy_usage'], label='Energy Usage')
plt.scatter(df[df['anomaly']==1].index, df[df['anomaly']==1]['energy_usage'],
            color='red', label='Anomaly', marker='x', s=80)
plt.title("Isolation Forest Anomaly Detection")
plt.xlabel("Time")
plt.ylabel("Energy Usage (MW)")
plt.legend()
plt.show()

# ------------------------------
# Step 4: Short-Term Forecasting with SARIMAX
# ------------------------------
model = SARIMAX(df['energy_usage'],
                exog=df[['temperature']],
                order=(1,1,1),
                seasonal_order=(1,1,1,96), # 96 periods in a day
                enforce_stationarity=False,
                enforce_invertibility=False)
results = model.fit(disp=False)

# Forecast next 12 periods (3 hours)
future_exog = df[['temperature']].iloc[-12:]
forecast = results.get_forecast(steps=12, exog=future_exog)
forecast_mean = forecast.predicted_mean
conf_int = forecast.conf_int()

# Plot forecast
plt.figure(figsize=(12,5))
plt.plot(df.index, df['energy_usage'], label='Observed')
plt.plot(forecast_mean.index, forecast_mean, color='red', label='Forecast')
plt.fill_between(conf_int.index, conf_int.iloc[:,0], conf_int.iloc[:,1], color='pink', alpha=0.3)
plt.title("SARIMAX Short-Term Forecast")
plt.xlabel("Time")
plt.ylabel("Energy Usage (MW)")
plt.legend()
plt.show()
```

Anomaly Detection Plot

Red crosses indicate spikes/drops in energy usage

Blue line shows normal energy consumption

SARIMAX Forecast Plot

Red line: short-term forecast for next 12 intervals

Pink shaded area: confidence interval
